```python
import torch

matrix=torch.randint(1,8,(2,3))

matrix

```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))

plt.imshow(matrix.numpy())
plt.axis('off')
plt.title('Random Matrix')
plt.colorbar()
plt.show()
```

```python
scalar_toplam=matrix+10

scalar_toplam
```

```python
scalar_cikarma=matrix-10

print('Cikarma:',scalar_cikarma)

scalar_carpma=matrix*10

print('Carpma:',scalar_carpma)

scalar_bolme=matrix/10

print('Bolme:',scalar_bolme)

scalar_ustu=matrix**10

print('Ustu:',scalar_ustu)
```
