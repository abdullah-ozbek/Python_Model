#### Numerik Türev

```python
import numpy as np

def f(x):
    return x**2 + 3*x + 2

x=2
h=1e-5
turev = (f(x+h)-f(x-h))/(2*h)

print(turev)
```

#### Sembolik Türev

```python
import sympy as sp

x = sp.symbols('x')
f = x**2 + 3*x + 2

turev = sp.diff(f,x)

print(turev)
```

```python
import torch

x=torch.tensor(1.0, requires_grad=True, dtype=torch.float32)

y=x**3 + 3*x + 5
z=y**2 + y
h=2*z

h.backward()

print('x=', x.item())
print('y=', y.item())
print('z=', z.item())
print('dh/dx=', x.grad.item())
```
